```python
import pandas as pd
import numpy as np
import os

# Read artifact from Notebook 1
artifact_in = os.path.join("artifacts", "ml_table.csv")
df = pd.read_csv(artifact_in)

print(f"Loaded ML Table shape: {df.shape}")
```

```python
# Convert delivery dates to datetime format
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])

# Filter only delivered orders for labeling
df_delivered = df[df['order_status'] == 'delivered'].copy()

# Create binary label: 1 if delivered after estimate, 0 otherwise
df_delivered['is_late'] = (
    df_delivered['order_delivered_customer_date'] > df_delivered['order_estimated_delivery_date']
).astype(int)

print(f"Delivered orders shape: {df_delivered.shape}")
```

```python
# Check label distribution
label_counts = df_delivered['is_late'].value_counts()
label_proportions = df_delivered['is_late'].value_counts(normalize=True) * 100

print("Label Distribution (Counts):")
print(label_counts)
print("\nLabel Distribution (Percentages):")
print(f"On-Time (0): {label_proportions.get(0, 0):.2f}%")
print(f"Late    (1): {label_proportions.get(1, 0):.2f}%")

# Sanity check on sample rows
sample_check = df_delivered[[
    'order_delivered_customer_date', 
    'order_estimated_delivery_date', 
    'is_late'
]].head(5)
print("\nSample Check:")
print(sample_check)
```

```python
# Save artifact for Notebook 3
artifact_out = os.path.join("artifacts", "labeled_table.csv")
df_delivered.to_csv(artifact_out, index=False)

print(f"Labeled Artifact saved successfully to: {artifact_out}")
```
