```python
import pandas as pd
import numpy as np
import os

# Read artifact from Notebook 2
artifact_in = os.path.join("artifacts", "labeled_table.csv")
df = pd.read_csv(artifact_in)

# Sort chronologically by purchase timestamp
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
df = df.sort_values('order_purchase_timestamp').reset_index(drop=True)

print(f"Total dataset shape: {df.shape}")
print(f"Date range: {df['order_purchase_timestamp'].min()} to {df['order_purchase_timestamp'].max()}")
```

```python
n = len(df)
train_end = int(n * 0.70)
val_end = int(n * 0.85)

train_df = df.iloc[:train_end].copy()
val_df = df.iloc[train_end:val_end].copy()
test_df = df.iloc[val_end:].copy()

print(f"Train shape: {train_df.shape} ({len(train_df)/n*100:.1f}%)")
print(f"Val shape:   {val_df.shape} ({len(val_df)/n*100:.1f}%)")
print(f"Test shape:  {test_df.shape} ({len(test_df)/n*100:.1f}%)")
```

```python
# Check date bounds for each split
print("--- Train Period ---")
print(f"{train_df['order_purchase_timestamp'].min()} -> {train_df['order_purchase_timestamp'].max()}")
print(f"Late Ratio: {train_df['is_late'].mean()*100:.2f}%\n")

print("--- Validation Period ---")
print(f"{val_df['order_purchase_timestamp'].min()} -> {val_df['order_purchase_timestamp'].max()}")
print(f"Late Ratio: {val_df['is_late'].mean()*100:.2f}%\n")

print("--- Test Period ---")
print(f"{test_df['order_purchase_timestamp'].min()} -> {test_df['order_purchase_timestamp'].max()}")
print(f"Late Ratio: {test_df['is_late'].mean()*100:.2f}%")
```

```python
# Save artifacts for Notebook 4 & 5
train_df.to_csv(os.path.join("artifacts", "train.csv"), index=False)
val_df.to_csv(os.path.join("artifacts", "val.csv"), index=False)
test_df.to_csv(os.path.join("artifacts", "test.csv"), index=False)

print("Split artifacts saved successfully: train.csv, val.csv, test.csv")
```
